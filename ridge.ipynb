# Ridge Regression

A simple toy example using ridge regression.

```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

from pai import RidgeRegression

mpl.rc_file_defaults()
```

## Toy Data

```python
rng = np.random.default_rng(42)

x_base = np.linspace(-10, 10, 100).reshape((100, 1))
x = rng.uniform(-10, 10, size=(100, 1))
y = x + rng.normal(size=(100, 1)) + 2
```

```python
fig, ax = plt.subplots(figsize=(4, 4))

ax.scatter(x, y, label="Toy Data", s=10)
ax.plot(x_base, x_base + 2, label="DGF", color="orange")
ax.legend()

plt.show()
```

## Fit Ridge Regression

The implementation uses mini-batch gradient descent to optimize the objective function.

```python
ridge = RidgeRegression(alpha=0.9)
ridge = ridge.fit(x, y)

pred_y = ridge.predict(x_base)
```

```python
fig, ax = plt.subplots(figsize=(4, 4))

ax.scatter(x_base, pred_y, label=f"Predictions", s=10)
ax.plot(x_base, x_base + 2, label="DGF", color="orange")
ax.legend()

plt.show()
```
